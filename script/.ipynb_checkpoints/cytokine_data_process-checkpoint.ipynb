```python
cytokine = ['IFN-a_144Nd', 'CXCL9_146Nd', 'CD86_147Sm','H3K27ac_148Nd', 
           'IL-1b_149Sm', '4E-BP1_150Nd', 'pS6_151Eu','CCR2_152Sm', 'TNF-a_153Eu', 
           'IL-4_154Sm',  'CASP-8_156Gd', 'CD38_159Tb', 
           'CD123_161Dy','Ki-67_162Dy', 'CXCL10_163Dy', 'Arginase-1_164Dy', 
           'IL-6_165Ho','MCP-1_166Er', 'CD69_167Er', 'STC1_168Er', 'GZMB_169Tm', 
           'HLA-DR_171Yb', 'IFN-g_172Yb', 'pSTAT1_175Lu', 'pSTAT3_176Yb']
```

```python
# =========================
# 1️⃣ 提取数据
# =========================
df = adata.to_df()[cytokine]
df["Participant"] = adata.obs["Participant"].values
df["Timepoint_V"] = adata.obs["Timepoint_V"].values
```

```python
# =========================
# 2️⃣ 计算中位数
# =========================
median_df = (
    df.groupby(["Participant", "Timepoint_V"], observed=False)
      .median()
      .reset_index()
)
# 转成长表
long_df = median_df.melt(
    id_vars=["Participant", "Timepoint_V"],
    value_vars=cytokine,
    var_name="cytokine",
    value_name="median_expr"
)
```

```python
# =========================
# 3️⃣ 显著性函数
# =========================
def get_star(p):
    if p < 1e-4:
        return "****"
    elif p < 1e-3:
        return "***"
    elif p < 1e-2:
        return "**"
    elif p < 0.05:
        return "*"
    else:
        return "ns"
```

```python
# =========================
# 4️⃣ 作图
# =========================
from scipy.stats import ttest_rel
for c in cytokine:
    sub = long_df[long_df["cytokine"] == c]

    plt.figure(figsize=(5, 5))
    ax = sns.boxplot(
        data=sub,
        x="Timepoint_V",
        y="median_expr"
    )
    sns.stripplot(
        data=sub,
        x="Timepoint_V",
        y="median_expr",
        color="black",
        size=4,
        alpha=0.6
    )

    # =========================
    # 5️⃣ 统计检验（配对）
    # =========================
    pivot = sub.pivot(
        index="Participant",
        columns="Timepoint_V",
        values="median_expr"
    )

    # 两组比较
    comparisons = [
        ("V1Day0", "V1Day1"),
        ("V2Day0", "V2Day1")
    ]

    y_max = sub["median_expr"].max()

    for i, (t1, t2) in enumerate(comparisons):
        if t1 in pivot.columns and t2 in pivot.columns:
            paired = pivot[[t1, t2]].dropna()

            if len(paired) > 0:
                '''stat, p = wilcoxon(paired[t1], paired[t2])'''
                stat, p = ttest_rel(paired[t1], paired[t2])

                star = get_star(p)

                # 标注位置
                x1, x2 = i*2, i*2+1
                y = y_max * 1.01 

                ax.plot([x1, x1, x2, x2], [y, y*1.005, y*1.005, y], lw=1.5)
                ax.text((x1+x2)/2, y*1.006, star, ha='center')

    plt.title(c)
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

adata_df = adata.to_df()[cytokine]

# 把分组信息加进来
adata_df["Timepoint_V"] = adata.obs["Timepoint_V"].values

# 4个时间点（可自动获取）
groups = adata_df["Timepoint_V"].unique()

# 颜色（可自定义）
colors = ["red", "blue", "green", "orange"]

plt.figure(figsize=(20, 20))

for i, col in enumerate(cytokine):
    plt.subplot(5, 5, i+1)
    
    # ⭐ 所有数据用于统一x轴
    all_data = adata_df[col].dropna()
    xmin = np.quantile(all_data, 0.01)
    
    # ⭐ 按 group 叠加画
    for g, c in zip(groups, colors):
        data = adata_df.loc[
            adata_df["Timepoint_V"] == g, col
        ].dropna()
        
        plt.hist(
            data,
            bins=500,        # ⭐ 建议少一点，不然太密
            alpha=0.3,       # ⭐ 透明度关键
            color=c,
            label=g,
            density=True     # ⭐ 推荐（可比较形状）
        )
    
    plt.xlim(xmin, data.max())
    plt.title(col)
    
    if i == 0:
        plt.legend()

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

adata_df = adata.to_df()[cytokine]
adata_df["Timepoint_V"] = adata.obs["Timepoint_V"].values

groups = adata_df["Timepoint_V"].unique()
colors = ["red", "blue", "green", "orange"]

plt.figure(figsize=(20, 20))

for i, col in enumerate(cytokine):
    plt.subplot(5, 5, i+1)
    
    all_data = adata_df[col].dropna()
    xmin = np.quantile(all_data, 0.01)
    xmax = np.quantile(all_data, 0.99)
    
    for g, c in zip(groups, colors):
        data = adata_df.loc[
            adata_df["Timepoint_V"] == g, col
        ].dropna()
        
        # ⭐ 1. 透明填充
        plt.hist(
            data,
            bins=300,
            alpha=0.3,
            color=c,
            density=True
        )
        
        # ⭐ 2. 不透明轮廓（关键）
        plt.hist(
            data,
            bins=300,
            histtype="step",   # ⭐ 只画边框
            linewidth=1.5,
            color=c,
            density=True
        )
    
    plt.xlim(xmin, data.max())
    plt.title(col)
    
    if i == 0:
        plt.legend(groups)

plt.tight_layout()
plt.show()
```
